```python
import pandas as pd

path = r"C:\Users\POP\Desktop\portofolio\projects\customer trends data analysis-SQL-Python-PowerBI\customer_shopping_behavior.csv"
df = pd.read_csv(path, encoding="utf-8-sig")
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby("Category")['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Create a column age_group
labels=['Young Adult ','Adult','Middle_aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'] = df['frequency_of_purchases'].replace({
    'Bi-Weekly': 'Fortnightly',
    'Every 3 Months': 'Quarterly'
})
```

```python
# Verify: exactly 5 categories now
df['frequency_of_purchases'].value_counts()
```

```python
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}

df['purchases_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df['purchases_frequency_days'].isna().sum()
```

```python
df[['purchases_frequency_days','frequency_of_purchases']].head(10)
```

```python
df['purchases_frequency_days'] = df['purchases_frequency_days'].astype('Int64')
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
df.groupby('frequency_of_purchases')['purchases_frequency_days'].unique()
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used',axis=1)
```
